While loop

```python
i=0
while(i<3):
    print("Hello World")
    i+=1
```

For loop

```python
for i in [0,0,0]:
    print("Hello World")
print("\n")
for i in range (3):
    print("Hello World")
print("\n")
i = [0,1,2]
for i in i:
    print("Hello World")
```
