```python
x = int(input("Enter your mark: "))

if x > 89 and x < 101:
    print("A+")
elif x > 79 and x < 90:
    print("A")
elif x > 69 and x < 80:
    print("B+")
elif x > 59 and x < 70:
    print("B")
else:
    print("Fail")
```

match similar to switch

```python
x = int(input("Enter your marks: "))

match x:
    case x if x > 89 and x < 101:
        print("A+")
    case x if x > 79 and x < 90:
        print("A")
    case x if x > 69 and x < 80:
        print("B+") 
    case x if x > 59 and x < 70:
        print("B")
    case _:
        print("Fail")
```

bitwise operators

```python
x = input("Enter your name: ")

match x:
    case "JP" | "jp":
        print("Hello JP")
    case "Ruby" | "ruby":
        print("Hello Honeywell")
    case _: 
        print("Hello User")
```
